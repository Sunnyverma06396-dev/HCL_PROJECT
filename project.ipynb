**Hour vs Bike Demand**

```python
hourly = df.groupby('Hour')['Rented Bike Count'].mean()
hourly
plt.figure(figsize=(10,5))
hourly.plot()
plt.title('Bike Demand by Hour')
plt.show()
```

**Finding**

*   Morning Pick-up Hours

*   Evening Return Hours



**Overall Dataset Inspection Conclusion**

The dataset was successfully loaded and validated. No major data quality issues were observed. Column types were verified and the dataset was suitable for exploratory data analysis and demand pattern investigation.


**Season Wise Demand**

```python
season = df.groupby('Seasons')['Rented Bike Count'].mean()

season.plot(kind='bar')
plt.title('Season Wise Bike Rentals')
plt.show()
```

**Finding**


*   Summer highest demand
*   Winter lowest demand

 **Missing Values & Data Cleaning Conclusion**

 Missing values and duplicate records were checked before analysis. Data cleaning ensured that the results were reliable and that no significant inconsistencies affected the interpretation of bike rental demand.

**Holiday vs Non-Holiday**

```python
holiday = df.groupby('Holiday')['Rented Bike Count'].mean()
holiday.plot(kind='bar')
plt.title('Holiday vs Non-Holiday')
plt.show()
```

 **Finding**

People generally rent more during working days.

**Hourly Bike Demand Analysis Conclusion**

Bike rentals varied significantly across different hours of the day. Demand was highest during commuter hours, indicating that bicycles are frequently used for daily transportation. Late-night and early-morning hours showed comparatively lower demand.

**Temperature effect**

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x='Temperature(°C)',y='Rented Bike Count',data=df,alpha=0.6)
plt.title('Temperature vs Bike Rentals')
plt.show()
```

**Finding**

Demand increases with moderate temperature.

**Season-wise Demand Conclusion**

Seasonal trends had a strong impact on bike rentals. Higher rental activity was observed during warmer seasons, while colder seasons experienced a noticeable decline in demand due to less favorable weather conditions.

**Rainfall Effect**

```python
sns.scatterplot(x='Rainfall(mm)',y='Rented Bike Count',data=df,alpha=0.6)
plt.title('Rainfall vs Bike Rentals')
plt.show()
```

Finding

Heavy rainfall reduces bike rentals.

**Holiday vs Non-holiday Analysis Conclusion**

Bike rental demand was generally higher on non-holidays compared to holidays. This suggests that bicycles are commonly used for commuting, education, and routine travel purposes during regular working days.

**Humidity Impact**

```python
sns.boxplot(
    x='Humidity(%)',
    y='Rented Bike Count',
    data=df
)
plt.show()
```

**Humidity Impact Conclusion**

Higher humidity levels were generally associated with reduced bike rental activity. Excessively humid conditions may decrease user comfort and consequently lower the demand for bicycles.

**Monthly Demand**

```python
monthly = df.groupby('Month')['Rented Bike Count'].mean()
monthly.plot(kind='bar')
plt.title('Monthly Bike Rentals')
plt.show()
```

**Monthly Demand Analysis Conclusion**

Monthly rental demand displayed clear seasonal fluctuations. Certain months consistently recorded higher bike usage, reflecting the influence of climate and outdoor activity patterns throughout the year.

**Correlation Matrix**

```python
numeric = df.select_dtypes(include=np.number)
plt.figure(figsize=(10,10))
sns.heatmap(numeric.corr(),annot=True)
cmap = 'Coolwarm'
plt.show()
```

**Correlation Analysis Conclusion**

Correlation analysis indicated that temperature showed a positive association with bike rental demand, while environmental factors such as rainfall, snowfall, and humidity demonstrated negative relationships. These findings highlight the importance of weather conditions in predicting bike-sharing demand.
